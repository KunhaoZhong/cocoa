```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
sn_data_file = './external_modules/data/sn_data/DESY5/DES-SN5YR_HD.csv'
sn_cov_file = './external_modules/data/sn_data/DESY5/covsys_000.txt'
```

```python
sn_data = pd.read_csv(sn_data_file)
print('number of total sn', sn_data.shape[0])
```

```python
# Mask data with z >= 0.1
mask = sn_data["zCMB"] >= 0.1
filtered_data = sn_data[mask].reset_index(drop=True)

with open(sn_cov_file, "r") as f:
    lines = f.readlines()
    n = int(lines[0])
    flat_cov = np.fromstring(' '.join(lines[1:]), sep=' ')
    covmat = flat_cov.reshape((n, n))

# Get the indices of the kept rows
kept_indices = mask.to_numpy().nonzero()[0]

# Slice the covariance matrix
filtered_cov = covmat[np.ix_(kept_indices, kept_indices)]

# Save the filtered data
filtered_data.to_csv("./external_modules/data/sn_data/DESY5_Z01/DES-SN5YR_HD.csv", index=False)

# Save the filtered covariance matrix (flattened, row-major, one per line)
with open("./external_modules/data/sn_data/DESY5_Z01/covsys_000.txt", "w") as f:
    new_n = len(filtered_data)
    f.write(f"{new_n}\n")
    for val in filtered_cov.flatten():
        f.write(f"{val:.12f}\n")

print('number of sn left', kept_indices.shape[0])
```
